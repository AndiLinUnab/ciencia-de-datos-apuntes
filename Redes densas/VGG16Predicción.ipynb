```python
import tensorflow as tf
from tensorflow.keras.applications.vgg16 import VGG16, preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image
import numpy as np
import requests
from PIL import Image
from io import BytesIO

# 1. Cargar el modelo VGG16 preentrenado en ImageNet
model = VGG16(weights='imagenet')

# 2. Descargar la imagen desde la URL
url = 'https://misanimales.com/wp-content/uploads/2021/05/hormiga-culona-cara.jpg'
response = requests.get(url)
img = Image.open(BytesIO(response.content))

# 3. Redimensionar a 224x224 y convertir a array
img_resized = img.resize((224, 224))
x = image.img_to_array(img_resized)
x = np.expand_dims(x, axis=0)
x = preprocess_input(x)

# 4. Realizar la predicción
preds = model.predict(x)

# 5. Decodificar las primeras 3 predicciones
decoded_preds = decode_predictions(preds, top=3)[0]

# 6. Mostrar los resultados en formato de porcentaje
print("Top 3 Predicciones:")
for i, (imagenet_id, label, prob) in enumerate(decoded_preds):
    percentage = prob * 100
    print(f"{i+1}. {label}: {percentage:.2f}%")

# 7. Imprimir la predicción con el porcentaje más alto
best_label, best_prob = decoded_preds[0][1], decoded_preds[0][2]
print(f"\nPredicción con mayor porcentaje: {best_label} ({best_prob * 100:.2f}%)")
```
